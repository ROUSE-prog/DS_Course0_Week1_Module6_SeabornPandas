# Data Visualizations with Seaborn & Pandas

## Business Context
This lab explores how visualization libraries can turn raw numerical and categorical data into clearer analytical stories. It combines Seaborn's high-level statistical graphics with pandas' built-in plotting tools.

The analysis progresses from basic boxplots and presentation styling to two real datasets: penguin biometrics and iris flower measurements.

## Objectives

- Label plots with appropriate axis labels and titles

## Part I: Seaborn

```python
# CodeGrade step0
# Run this cell without changes
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline

seed = 42
np.random.seed(seed)
data = np.random.normal(size=(20, 10)) + np.arange(10) / 2
```

### Step 1 — Create a Basic Boxplot
Create a boxplot and save the returned Axes object as `boxplot1`.

```python
# CodeGrade step1
boxplot1 = sns.boxplot(data=data)
```

### Analyst Note
A boxplot summarizes the center, spread, and potential outliers for each numerical column in a compact form.

### Step 2 — Add Labels and a Title
Repeat the boxplot and label it with the required axis labels and title.

```python
# CodeGrade step2
boxplot2 = sns.boxplot(data=data)
boxplot2.set(xlabel='X Label', ylabel='Y Label', title='Example Boxplot')
```

### Step 3 — Apply the Darkgrid Style
Use Seaborn's `darkgrid` style, then recreate the labeled boxplot.

```python
# CodeGrade step3
sns.set_style('darkgrid')
boxplot3 = sns.boxplot(data=data)
boxplot3.set(xlabel='X Label', ylabel='Y Label', title='Example Boxplot')
```

### Step 4 — Use a Presentation Context
Set the context to `poster` so fonts and plot elements are scaled for presentation use.

```python
# CodeGrade step4
sns.set_context('poster')
boxplot4 = sns.boxplot(data=data)
boxplot4.set(xlabel='X Label', ylabel='Y Label', title='Example Boxplot')
```

## Penguin Dataset
The next two visualizations use biometric data for three penguin species.

```python
# CodeGrade step0
# Run this cell without changes
penguins = pd.read_csv('penguins.csv')
penguins.head()
```

### Step 5 — Penguin Body Mass Distribution
Create a histogram of `body_mass_g`, use color to distinguish sex, set context to `talk`, and add the required labels and title.

```python
# CodeGrade step5
sns.set_context('talk')
histplot1 = sns.histplot(data=penguins, x='body_mass_g', hue='sex')
histplot1.set(xlabel='Body Mass (g)', ylabel='Number of Penguins', title='Penguin Mass Distribution by Sex')
```

### Interpretation
The histogram makes it possible to compare the body-mass distributions of male and female penguins directly rather than looking only at individual observations.

### Step 6 — Penguin Bill Size Scatterplot
Plot bill length against bill depth, use color for sex, marker style for species, set context to `paper`, and add the required labels and title.

```python
# CodeGrade step6
sns.set_context('paper')
scatterplot1 = sns.scatterplot(data=penguins, x='bill_length_mm', y='bill_depth_mm', hue='sex', style='species')
scatterplot1.set(xlabel='Bill Length (mm)', ylabel='Bill Depth (mm)', title='Scatterplot of Penguin Bill Sizes')
```

### Interpretation
The scatterplot shows that bill dimensions vary substantially across species, while sex adds a second categorical comparison through color.

## Part II: Pandas

The final section uses pandas plotting on the classic iris dataset.

```python
# CodeGrade step0
# Run this cell without changes
iris = pd.read_csv('iris.csv')
iris.head()
```

### Step 7 — Mean Sepal Width by Species
Group by species, calculate the mean sepal width, and plot the resulting Series as a bar chart.

```python
# CodeGrade step7
species_mean_sepal_width = iris.groupby('species').mean(numeric_only=True)['sepal_width']
barplot1 = species_mean_sepal_width.plot(kind='bar', title='Distribution of Sepal Width', xlabel='Species', ylabel='Mean Sepal Width')
```

### Interpretation
The grouped bar chart provides a direct species-level comparison of average sepal width.

### Step 8 — Sepal Length Distributions by Species
Use pandas `.plot()` to create three histograms, one for each species, in a single call.

```python
# CodeGrade step8
histplot2 = iris.hist(column='sepal_length', by='species')
```
